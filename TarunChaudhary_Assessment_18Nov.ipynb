```python
# Q3
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

X, y = load_wine(return_X_y=True, as_frame=True)


# (a) Print shape of X and count of each class in y

print("Shape of X:", X.shape)

print("\nCount of each class:")
print(y.value_counts())


# (b) Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# (c) Scale the features using StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


# (d) Train Logistic Regression model

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

# Make predictions
y_pred = model.predict(X_test_scaled)

# Print test accuracy
print("\nTest Accuracy:", accuracy_score(y_test, y_pred))

# Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

Precision for class 0 means, out of all the samples predicted by the model as class 0, how many actually belong to class 0.

For example, if the model predicts 20 samples as class 0 and 18 are actually class 0, precision is:0.90

```python
# Q2
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import pandas as pd


# (a) Train Decision Tree Classifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

dt_pred = dt_model.predict(X_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, dt_pred))


# (b) Train Random Forest Classifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("\nRandom Forest Accuracy:", accuracy_score(y_test, rf_pred))


# (c) Print top 3 features by importance

importances = rf_model.feature_importances_

feature_importance = pd.Series(importances, index=X_train.columns)

top_3 = feature_importance.sort_values(ascending=False).head(3)

print("\nTop 3 Important Features:")
print(top_3.to_string())
```

Random Forest usually performs better because it combines predictions from multiple decision trees, reducing overfitting and improving prediction accuracy compared to a single tree.

```python
# Q3
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


# (a) KNN on unscaled data

knn_unscaled = KNeighborsClassifier(n_neighbors=5)

knn_unscaled.fit(X_train, y_train)

pred_unscaled = knn_unscaled.predict(X_test)

accuracy_unscaled = accuracy_score(y_test, pred_unscaled)

print("KNN Accuracy (Unscaled):", accuracy_unscaled)


# (b) KNN on scaled data

knn_scaled = KNeighborsClassifier(n_neighbors=5)

knn_scaled.fit(X_train_scaled, y_train)

pred_scaled = knn_scaled.predict(X_test_scaled)

accuracy_scaled = accuracy_score(y_test, pred_scaled)

print("KNN Accuracy (Scaled):", accuracy_scaled)


# (c) Print both accuracies on one line

print("\nUnscaled Accuracy:", accuracy_unscaled,
      "| Scaled Accuracy:", accuracy_scaled)
```

Scaling changes KNN results significantly because KNN calculates distances between data points, so features with larger numerical ranges can dominate the distance calculation, whereas Decision Trees use feature-based splitting rather than distance calculations and are therefore generally unaffected by scaling.

```python
# Q4
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import pandas as pd


# (a) Scale all features of X

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# (b) Apply K-Means clustering

kmeans = KMeans(
    n_clusters=3,
    n_init=10,
    random_state=42
)

cluster_labels = kmeans.fit_predict(X_scaled)


# (c) Count the number of rows in each cluster

print("Number of rows in each cluster:")
print(pd.Series(cluster_labels).value_counts().sort_index())


# (d) Attach cluster labels to a DataFrame

cluster_df = X.copy()

cluster_df["cluster"] = cluster_labels

# Calculate mean of selected features for each cluster

print("\nMean values for each cluster:")

print(
    cluster_df.groupby("cluster")[
        ["alcohol", "proline", "color_intensity"]
    ].mean()
)
```

Cluster 0 -Low-alcohol, low-proline and light-coloured wines
Cluster 1- Moderate-alcohol, moderate-proline and dark-coloured wines
Cluster 2- High-alcohol, high-proline and moderately coloured wines

```python
#Q5
import csv, json

data = [
    ["name", "marks", "city"],
    ["Asha", 78, "Bangalore"],
    ["Ravi", "", "Chennai"],
    ["Meena", 91, "Bangalore"],
    ["Imran", 64, "Hyderabad"],
    ["Divya", 88, ""]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

def read_marks(path):

    # (a) Open file using try-except
    try:
        with open(path, "r") as file:

            # (b) Read rows using DictReader
            reader = csv.DictReader(file)

            total_rows = 0
            count_marks = 0
            total_marks = 0

            for row in reader:

                total_rows += 1

                # (c) Skip rows with empty marks
                if row["marks"] == "":
                    continue

                count_marks += 1
                total_marks += int(row["marks"])

            if count_marks > 0:
                average = total_marks / count_marks
            else:
                average = 0

            summary = {
                "total_rows": total_rows,
                "rows_with_marks": count_marks,
                "average_marks": average
            }

            # (d) Write dictionary to JSON file
            with open("summary.json", "w") as json_file:
                json.dump(summary, json_file, indent=4)

            print("\nSummary dictionary:")
            print(summary)

            print("\nContents of summary.json:")

            with open("summary.json", "r") as json_file:
                print(json_file.read())

            return summary

    except FileNotFoundError:
        print(f"File not found: {path}")
        return None


# Call the function twice

print("First function call:")
read_marks("students.csv")

print("\nSecond function call:")
read_marks("missing.csv")
```
